```python
from src.signals.utils import get_wiki_adjacency
from src.generation.utils import get_tokenizer

MODEL_NAME = "meta-llama/Llama-3.2-3B-Instruct"
tokenizer = get_tokenizer(MODEL_NAME)

def get_similarity(word_1, word_2):
    tok_1 = tokenizer.encode(word_1, add_special_tokens=False)[0]
    tok_2 = tokenizer.encode(word_2, add_special_tokens=False)[0]

    wiki_adjacency = get_wiki_adjacency(
        [tok_1, tok_2],
        nu=8,
        model_name=MODEL_NAME,
        prefix_entail=True,
        # lang="en",
    )
    return wiki_adjacency[1,0]

```

```python
get_similarity("student", "school")
```

```python
import numpy as np

synonyms_scores = [get_similarity(*syns) for syns in synonyms]
hypernyms_scores = [get_similarity(*syns) for syns in hypernyms]
cooccurring_scores = [get_similarity(*syns) for syns in cooccurring]

synonyms_similarity = np.mean(synonyms_scores)
hypernyms_similarity = np.mean(hypernyms_scores)
cooccurring_similarity = np.mean(cooccurring_scores)

synonyms_std = np.std(synonyms_scores)
hypernyms_std = np.std(hypernyms_scores)
cooccurring_std = np.std(cooccurring_scores)

# contradicting_adjectives_similarity =  np.mean([get_similarity(*syns) for syns in contradicting_adjectives])
# adjective_synonyms_similarity =  np.mean([get_similarity(*syns) for syns in adjective_synonyms])


print(f"synonyms_similarity = {synonyms_similarity:.4f} +/- {synonyms_std:.4f}")
print(f"hypernyms_similarity = {hypernyms_similarity:.4f} +/- {hypernyms_std:.4f}")
print(f"cooccurring_similarity = {cooccurring_similarity:.4f} +/- {cooccurring_std:.4f}")
# print(f"contradicting_adjectives_similarity = {contradicting_adjectives_similarity}")
# print(f"adjective_synonyms_similarity = {adjective_synonyms_similarity}")
```

```python
# Split cooccurring into 10 categories of 10 pairs each
categories = {
    "European countries": cooccurring[:10],
    "Social media platforms": cooccurring[10:20],
    "Months": cooccurring[20:30],
    "Computer hardware": cooccurring[30:40],
    "Animals": cooccurring[40:50],
    "French cities": cooccurring[50:60],
    "Car models": cooccurring[60:70],
    "Programming languages": cooccurring[70:80],
    "Computer hardwares": cooccurring[80:90],
    "Cocktails": cooccurring[90:100],
}

# Calculate mean and std for each category
for name, pairs in categories.items():
    scores = [get_similarity(*pair) for pair in pairs]
    mean_val = np.mean(scores)
    std_val = np.std(scores)
    print(f"{name}: {mean_val:.4f} +/- {std_val:.4f}")
```
